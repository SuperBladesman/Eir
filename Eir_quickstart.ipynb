## One structure

`eir.check()` returns the verdict on one CIF. Displayed on its own in a notebook, the result is the one-page report.

````python
r = eir.check('examples/MgO_wrong_cell.cif')
r
````

````python
print(r.verdict())
````

````python
r.tier, r.code, r.reason, r.evidence
````

## Rietveld output

Most refinement programs write element symbols without charges. Eir does not guess oxidation states: give them, or put a `config.json` beside the CIF.

````python
r = eir.check('examples/SrTiO3_refined.cif',
              oxidation_states={'Sr': 2, 'Ti': 4, 'O': -2})
print(r.verdict())
r.quote          # the index to quote, its value and its sigma
````

````python
r.write_report('SrTiO3_refined.eir.html')   # a page to keep or send
````

## A folder

`eir.check_folder()` runs the notebook's full pipeline on every CIF in a folder and writes the same `outputs/` folder inside it. Set `FOLDER` to your own.

````python
FOLDER = 'examples'
f = eir.check_folder(FOLDER, verbose=False,
                     oxidation_states={'Mg': 2, 'Sr': 2, 'Ti': 4, 'O': -2})
f
````

````python
f.structures[['filename', 'chemical_formula', 'structure_pgrm_tier',
              'structure_pgrm_code', 'gii_recommended', 'gii_recommended_value']]
````

````python
print(f.summary[:3000])
````
